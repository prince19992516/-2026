## 人工初始推导与预期

### 5.1 非线性方程

**预期 1（二分法迭代次数）** 对 $f(x)=x^2-2$ 在 $[1,2]$ 上二分，精度 $\varepsilon=0.01$。由
$|c_n-\xi|\le (b-a)/2^{n+1}$，需 $2^{n+1}\ge 100$，即 $n\ge \lceil\log_2 100\rceil-1=6$。实际第 7 次折半后区间长 $0.0078<0.01$，与 A1 表格一致。预期：第 0 步中点 1.5，第 1 步 1.25，第 2 步 1.375。

**预期 2（残差与根误差）** 取 $f(x)=(x-1)^{10}$，$x=1.1$。残差 $10^{-10}$，根误差 $0.1$。预期：仅凭残差小不能判根准；A1 与 A2 都应在这一点上一致。

**预期 3（不动点三种改写）** 对 $x^2=2$：
- $g_1=x^2+x-2$，$g_1'(\sqrt2)=2\sqrt2+1\approx3.83>1$，发散；
- $g_2=2/x$，$|g_2'(\sqrt2)|=1$，临界，从 1 出发在 1 与 2 间振荡；
- $g_3=x/2+1/x$，$g_3'(\sqrt2)=0$，超线性，且恰为 Newton 迭代。
预期：A1 与 A2 都应给出“改写不同、命运不同”的结论。

**预期 4（Newton 二次收敛常数）** 对 $x^2-2$，简单根 $\sqrt2$，$f''/(2f')=2/(2\cdot2\sqrt2)=1/(2\sqrt2)\approx0.3536$。预期：$e_{k+1}/e_k^2\to0.3536$。

**预期 5（重根退化）** 对 $f=(x-1)^2$，Newton 误差比 $e_{k+1}/e_k\to1/2$。预期：A1 的“多重根降为线性”结论正确；A2 的命题 5.1.8 给出同一常数 $(m-1)/m$。

**预期 6（Newton 二周期反例）** 对 $f=x^3-2x+2$，$x_0=0$：$f(0)=2,f'(0)=-2\Rightarrow x_1=1$；$f(1)=1,f'(1)=1\Rightarrow x_2=0$。预期：精确迭代在 0、1 间循环；A2 例 5.1.9 应给出同一反例。

### 5.2 下降方向

**预期 7（驻点三分类）** $f=x^2-y^2$ 在原点：梯度为零，Hessian $\mathrm{diag}(2,-2)$ 不定，鞍点。预期：A1 与 A2 都应判鞍点，且都应提醒“梯度为零≠极小”。

**预期 8（负梯度最速性）** 单位球约束下 $\min g^{\top}p$ 的解为 $p=-g/\|g\|$，由 Cauchy–Schwarz 等号条件给出。预期：两版都给出该结论。

**预期 9（步长上界）** 对 $\phi=\tfrac12 x^{\top}Ax$，固定步长梯度下降收敛当且仅当 $0<\alpha<2/\lambda_{\max}$。对 $A=\mathrm{diag}(2,8)$，$2/L=0.25$。预期：$\alpha=0.3$ 发散，$\alpha=0.1$ 收敛但慢；A1 图 5-13 图注若写“$2/L=0.1$”则与 $L=8$ 不符，属应核查项。

**预期 10（Armijo 回溯）** 对 $\phi=(x_1^2+20x_2^2)/2$，$x=(2,2)$，$p=-g=(-2,-40)$，$c_1=10^{-4}$，$\alpha_0=1$，$\beta=1/2$。手算 $\phi(x+\alpha p)=42-1604\alpha+16002\alpha^2$，Armijo 右端 $42-0.1604\alpha$。预期：$\alpha=1,1/2,1/4,1/8$ 均拒绝，$\alpha=1/16$ 接受，目标 $4.2578125$。A2 数值例 5.2.10 给出同一张表，可逐行核对。

**预期 11（Wolfe 曲率保障）** 若 $B_k\succ0$、$p_k=-B_k^{-1}g_k$、Wolfe 条件成立，则 $s_k^{\top}y_k\ge(c_2-1)g_k^{\top}s_k>0$。预期：A2 命题 5.2.15 与推论 5.3.8 应一致；A1 的 BFGS 章节应给出同一曲率条件。

### 5.3 Newton 与拟 Newton

**预期 12（求根 Newton vs 优化 Newton）** 对 $f=x^2-2$：求根 Newton 收敛到 $\sqrt2$；优化 Newton 解 $f'=2x=0$，一步到 0。预期：两版都应明确“公式同构、目标不同”。

**预期 13（Hessian 不定时方向非下降）** 对 $\phi=\tfrac12x_1^2-\tfrac12x_2^2$，$x=(0,1)$，$g=(0,-1)$，$H=\mathrm{diag}(1,-1)$，$p=(0,-1)$，$g^{\top}p=1>0$。预期：A2 例 5.3.3 应给出该反例；A1 例 5.3.1.2 给的是 $g^{\top}d=0$ 的临界反例，二者不矛盾但强度不同。

**预期 14（BFGS 一步）** $B_0=I$，$s=(1,0)^{\top}$，$y=(2,1)^{\top}$。手算 $B_1=\begin{pmatrix}2&1\\1&1.5\end{pmatrix}$，$B_1s=(2,1)^{\top}=y$，顺序主子式 2 与 2 均正。预期：A1 例 5.3.3.1 与 A2 例 5.3.10 应给出同一矩阵。

**预期 15（BFGS 正定保持）** 若 $B_k\succ0$ 且 $s^{\top}y>0$，则 $B_{k+1}\succ0$。证明用 $B$ 内积下的 Cauchy–Schwarz 与“两项同时为零则矛盾”。预期：A2 命题 5.3.7 给出完整证明；A1 只给思路，属“证明详略差异”，不是错误。

**预期 16（Armijo 不足以保正曲率）** 对 $\phi=-x^2/2$，$x=1$，$p=1$，任意 $\alpha>0$ 满足 Armijo，但 $s^{\top}y=-\alpha^2<0$。预期：A2 例 5.3.9 应给出该反例；A1 的“曲率条件不满足则跳过更新”与之一致。

### 5.4 非线性最小二乘

**预期 17（GN 线性一步）** 数据 $(0,1),(1,2),(2,4)$，模型 $a+bt$，从 $(0,0)$ 出发。若残差定义为 $r_i=y_i-(a+bt_i)$，Jacobian 取 $(-1,-t_i)$，则 $J^{\top}J=\begin{pmatrix}3&3\\3&5\end{pmatrix}$，$-J^{\top}r=(7,10)^{\top}$，解出 $(5/6,3/2)$，与 5.4.1 正规方程一致。预期：A1 例 5.4.2.1 的“先算 $-5/6$ 再用负号 Jacobian 重算”属于叙述绕路，应改为一开始就统一符号；A2 的式 (5.4.5) 用 $r=h-y$，符号相反但自洽，需在对照时说明“两版残差定义方向不同，结论一致”。

**预期 18（GN 完整步失败）** 对 $r(x)=x^2+1$，$x=0.1$，GN 步 $p=-(x^2+1)/(2x)=-5.05$，新点 $-4.95$，目标从 $0.505$ 增到约 $12.75$。预期：A2 例 5.4.6 应给出该反例；A1 的“GN 步可能过大，需线搜索或阻尼”与之一致。

**预期 19（LM 一步手算）** 同预期 17 的数据，$\lambda=1$。$J^{\top}J+I=\begin{pmatrix}4&3\\3&6\end{pmatrix}$，右端 $(7,10)^{\top}$，解 $d=(0.8,1.2667)$。预期：A1 例 5.4.3.1 给出 $(0.8,19/15)$，可逐项核对。

**预期 20（LM 大阻尼模型吻合）** 固定非驻点，$\lambda\to\infty$ 时 $p=O(1/\lambda)$，$\mathrm{pred}=\|g\|^2/\lambda+O(\lambda^{-2})$，$\mathrm{ared}-\mathrm{pred}=O(\lambda^{-2})$，故 $\rho\to1$。预期：A2 命题 5.4.8 给出该证明；A1 的“$\lambda$ 大像梯度下降、$\rho$ 大减 $\lambda$”与之一致。

**预期 21（指数模型局部辨识）** $h=ae^{-bt}$，$a\ne0$，至少两个不同 $t_i$，则 $J$ 满列秩；若所有 $t_i=t_0$，只识别组合 $ae^{-bt_0}$。预期：A2 命题 5.4.9 给出局部下界；A1 的“秩亏需加阻尼或检查可辨识性”与之一致。
