### 2026-11-15 第 1 条：核对“机器精度”符号口径
- **提示词（原话）**：
  > “A1 正文用 $u=2^{-p}$、$\epsilon_{\mathrm{mach}}=2^{1-p}$；A2 正文用 $\varepsilon=2^{1-p}$、$u=2^{-p}$。NumPy 的 `finfo.eps` 指哪一个？两稿是否数学等价？请给出判断依据，不要改写教材。”
- **AI 真实回复（原话摘录）**：
  > “NumPy `finfo.eps` 返回的是 1 与下一个可表示浮点数之间的间距，即 $2^{1-p}$。A1 的 $\epsilon_{\mathrm{mach}}$ 与 A2 的 $\varepsilon$ 是同一个量；A1 的 $u$ 与 A2 的 $u$ 也都是 $2^{-p}$。两稿数学等价，只是符号命名不同。”
- **本组人工核验**：
  - 运行 `np.finfo(np.float64).eps` 得 $2.220446049250313\times10^{-16}=2^{-52}$，`np.finfo(np.float32).eps` 得 $1.1920928955078125\times10^{-7}=2^{-23}$。
  - 对照 A1 式 (1.6) 与 A2 1.1 节表格，确认二者分别称该量为 $\epsilon_{\mathrm{mach}}$ 与 $\varepsilon$，而 $u$ 均定义为 $2^{-p}$。
  - 查阅 NumPy 2.3 `finfo` 文档（A1 参考文献 [5]）确认 `eps` 定义。
- **采纳/排除理由**：AI 回复与人工运行、文档一致，**采纳**为“符号差异、数学等价”的判断。但本组不采纳其“可直接统一”的建议，因为教材读者可能把 $u$ 与 `eps` 混用；建议在综合稿中显式写“本章 $u=2^{-p}$，`finfo.eps`$=\epsilon_{\mathrm{mach}}=2^{1-p}$”，并在首次出现处加脚注。
- **本组判断（与 AI 原话分开）**：AI 的事实陈述正确；整合口径的编辑决定由本组作出，不由 AI 代判。

### 2026-11-15 第 2 条：复核开场问题的条件数推导
- **提示词（原话）**：
  > “请只根据 $d(x)=\sqrt{x+1}-\sqrt{x}=1/(\sqrt{x+1}+\sqrt{x})$，推导 $\kappa_d=|xd'/d|$，并说明 $x=10^{16}$ 时直接式误差量级为何可达 $O(ux)$。不要引用 A1 的结论，给出中间步骤。”
- **AI 真实回复（原话摘录）**：
  > “由 $d=1/(\sqrt{x+1}+\sqrt{x})$ 求导得 $d'=-d/(2\sqrt{x(x+1)})$，故 $\kappa_d=\frac12\sqrt{x/(x+1)}<1/2$。直接式中两次开方的绝对误差约 $u\sqrt x$，相减后除以 $d\sim1/(2\sqrt x)$，相对误差约 $2u x$，量级 $O(ux)$。”
- **本组人工核验**：
  - 手推：对 $d=1/(\sqrt{x+1}+\sqrt{x})$ 用商法则，得 $d'=-d/(2\sqrt{x(x+1)})$；代入 $\kappa_d=|xd'/d|$ 得 $\frac12\sqrt{x/(x+1)}$。与 AI 一致。
  - 手推误差量级：设两次开方相对误差 $\le u$，绝对误差 $\le u\sqrt{x+1}$、$\le u\sqrt x$；差值误差 $\le u(\sqrt{x+1}+\sqrt x)\approx2u\sqrt x$；除以 $d\approx1/(2\sqrt x)$，相对误差 $\approx4ux$。AI 写 $2ux$，常数因子不同，量级 $O(ux)$ 一致。
  - 数值：$x=10^{16}$，$u=1.11\times10^{-16}$，$ux\approx1.11$，与“相对误差可达 1”相符。
- **采纳/排除理由**：**采纳**量级结论；**修正**常数因子表述，本组在综合稿中写“$O(ux)$”而不写具体常数，避免与 A1 式 (1.20) 的 $O(ux)$ 冲突。
- **本组判断**：AI 推导可用，但常数未核；本组手推确认 $4ux$ 与 $2ux$ 同量级，不影响结论。

### 2026-11-15 第 3 条：检验“积分递推反向一定更准”的候选主张
- **提示词（原话）**：
  > “A1 例 1.5 中反向递推从 $I_9$ 反推到 $I_0$ 误差缩小。能否说‘反向递推总比正向递推更准’？请给出反例或条件。”
- **AI 真实回复（原话摘录）**：
  > “不能一概而论。反向递推的误差传播因子是 $1/n$，正向是 $n$；但若末值 $I_N$ 本身误差很大，反推到低阶仍可能不准。此外，若递推方向改变后新增舍入在每一步被放大，也可能抵消缩小效果。需要具体分析误差传播因子和末值精度。”
- **本组人工核验**：
  - 由 $\delta_n=(-1)^{N-n}\frac{n!}{N!}\delta_N$，反推到 $n=0$ 的因子是 $1/N!$，确实缩小；但若 $|\delta_N|$ 大于 $N!$ 倍目标误差，低阶仍不达标。
  - 构造教学反例（人工构造，非 AI 回复）：取 $N=9$、$\widehat I_9=0.5$（远超上界 0.1），反推 $I_0$ 误差约 $0.5/9!\approx1.38\times10^{-6}$，仍大于 $10^{-7}$。说明“反向更准”需末值在合理范围内。
  - 核对 A1 例 1.5 的末值 $0.0684$ 落在 $(0.0368,0.1)$ 内，故反推有效。
- **采纳/排除理由**：AI 的“不能一概而论”**采纳**；其“新增舍入可能抵消”属于定性提示，本组未构造出具体反例，标为“条件不足，暂不写入教材”。
- **本组判断**：教材应写“反向递推的误差传播因子为 $1/n$，但末值精度仍需与目标误差匹配”，不写“反向总更准”。

### 2026-11-15 第 4 条：Softmax 平移后尾概率为零的损失路径
- **提示词（原话）**：
  > “$z=(0,-1000)$，真实类别为第 2 类。binary64 中平移 Softmax 输出 $(1,0)$。先概率再取负对数得什么？由 logits 直接算交叉熵得什么？哪个更可靠？”
- **AI 真实回复（原话摘录）**：
  > “先概率再取负对数：$p_2=0$，$-\log0=+\infty$。直接由 logits 算：$\ell=\log(1+e^{-1000})-(z_2-m)\approx0-(-1000)=1000$。后者更可靠，因为避免了先存零概率再取对数。”
- **本组人工核验**：
  - 高精度参考：尾概率约 $5.076\times10^{-435}$，低于 binary64 最小正非规格化数 $2^{-1074}\approx4.94\times10^{-324}$，故存 0。
  - 直接路径：$m=0$，$\log(1+e^{-1000})\approx e^{-1000}$（远小于 $u$），在当前精度下和为 1，对数 0，$\ell\approx1000$。与 A1 表 1.14 一致。
  - 检查“把零概率截到小常数”的替代：会改变损失函数，不采纳。
- **采纳/排除理由**：**采纳** AI 的路径比较；本组补充“准确损失为 $1000+\log(1+e^{-1000})$，当前精度下即 1000”的表述，写入综合稿。
- **本组判断**：AI 回复与人工核验一致；教材应强调“概率全有限且和为 1 仍可能无法计算损失”，不写“Softmax 稳定所以损失一定可靠”。
