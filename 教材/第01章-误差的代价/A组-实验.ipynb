## 1. 学习目标
区分绝对误差与相对误差；解释问题条件数与算法稳定性的差别；识别相消误差。

## 2. 数学背景与适用范围
对真值 $x$ 与近似值 $\hat{x}$，绝对误差为 $|\hat{x}-x|$；当 $x\ne0$ 时，相对误差为 $|\hat{x}-x|/|x|$。相对误差在真值为零时没有定义，不能用随意加一个小数的方式掩盖这个问题。

对 $f(x)=\sqrt{x+1}-\sqrt{x}$，有等价式 $f(x)=1/(\sqrt{x+1}+\sqrt{x})$。前式在大 $x$ 时相消，后式避免两个接近数相减。本例说明代数等价不代表浮点实现同样可靠。条件数刻画问题对输入扰动的敏感程度，稳定性刻画算法的舍入误差传播，二者不能混用。

## 3. 实现与可执行断言
代码使用合成输入或解析函数，无需下载数据。断言检测上述具体示例，不构成一般正确性证明。


```python
import math
from decimal import Decimal, localcontext
import numpy as np

def stable_difference(x):
    if not math.isfinite(x) or x < 0:
        raise ValueError("x must be finite and nonnegative")
    return 1.0 / (math.sqrt(x + 1.0) + math.sqrt(x))

x = 1e16
naive = math.sqrt(x + 1.0) - math.sqrt(x)
stable = stable_difference(x)
with localcontext() as ctx:
    ctx.prec = 60
    xd = Decimal('1e16')
    reference = float((xd + 1).sqrt() - xd.sqrt())
assert naive == 0.0
assert abs(stable - reference) / reference < 1e-14
assert stable_difference(0.0) == 1.0
print({"naive": naive, "stable": stable, "reference": reference})
```
