## 6.1 PINN：原理与实现（2学时）

对一维 Poisson 方程求场 $u$，分别测量内部方程残差、边界误差和解误差。给候选解加上一个常数，哪项指标能察觉？

### 内部残差、边界误差与解误差

```python
import numpy as np

x = np.linspace(0, 1, 101)
exact = np.sin(np.pi*x)
candidate = exact + 3*x + 2
candidate_second_derivative = -np.pi**2*np.sin(np.pi*x)
forcing = np.pi**2*np.sin(np.pi*x)
residual = -candidate_second_derivative - forcing
boundary_error = np.max(np.abs(candidate[[0, -1]]))
solution_error = np.max(np.abs(candidate-exact))
assert np.max(np.abs(residual)) < 1e-12
assert boundary_error > 4.9
assert solution_error > 4.9
print({"interior_residual": float(np.max(np.abs(residual))),
       "boundary_error": float(boundary_error), "solution_error": float(solution_error)})
```
