```python
# 单元 1 · 6.1 / A1：同一 FD 网格的两套评价泛函 + 反例常数
# 代码来源：AI 生成初稿（DeepSeek Harness / deepseek-flash，2026-11-16）；组员核对 fd()/rel_61()/rel_64() 三处口径
import numpy as np
PI = np.pi
trap = getattr(np, 'trapezoid', np.trapz)
exact = lambda x: np.sin(PI * x)
forcing = lambda x: PI ** 2 * np.sin(PI * x)

def quadrature(cells=800, order=4):          # 6.4 的共同积分点：800 段 x 4 点 Gauss
    z, w = np.polynomial.legendre.leggauss(order)
    x = (np.arange(cells)[:, None] + (z[None, :] + 1) / 2) / cells
    return x.ravel(), np.broadcast_to(w[None, :] / (2 * cells), x.shape).ravel()
XQ, WQ = quadrature()
print('积分点 %d 个，权重和 %.15f，int sin^2 = %.15f' % (len(XQ), WQ.sum(), np.dot(WQ, exact(XQ) ** 2)))

def fd(n):
    h = 1 / (n - 1); x = np.linspace(0, 1, n); u = np.zeros(n); m = n - 2
    A = (np.diag(np.full(m, 2 / h ** 2)) + np.diag(np.full(m - 1, -1 / h ** 2), 1)
         + np.diag(np.full(m - 1, -1 / h ** 2), -1))
    u[1:-1] = np.linalg.solve(A, forcing(x[1:-1]))
    return x, u

def rel_61(x, u):   # 6.1 口径：节点上的未加权离散 L2 比（= A1组-实验-6.1.ipynb 的 fd_solve.rel）
    ue = exact(x); return float(np.linalg.norm(u - ue) / np.linalg.norm(ue))

def rel_64(x, u):   # 6.4 口径：分段线性重建 + 3200 内点（= A1组-实验-6.4.ipynb 的 metrics.rel_l2）
    v = np.interp(XQ, x, u)
    return float(np.sqrt(np.dot(WQ, (v - exact(XQ)) ** 2) / np.dot(WQ, exact(XQ) ** 2)))

print(f"{'N':>5}{'c_h-1(解析)':>16}{'max|e|':>14}{'6.1口径':>14}{'6.4口径':>14}{'比值':>9}")
rows = []
for n in [51, 101, 201, 401, 801]:
    x, u = fd(n); h = 1 / (n - 1)
    ch = PI ** 2 / (4 / h ** 2 * np.sin(PI * h / 2) ** 2) - 1
    rows.append((n, ch, float(np.abs(u - exact(x)).max()), rel_61(x, u), rel_64(x, u)))
    print(f"{n:5d}{ch:16.6e}{rows[-1][2]:14.6e}{rows[-1][3]:14.6e}{rows[-1][4]:14.6e}{rows[-1][3]/rows[-1][4]:9.5f}")
print('1/sqrt(5) = %.10f ; max|e| 与 (c_h-1) 的最大差 = %.3e'
      % (1 / np.sqrt(5), max(abs(r[2] - r[1]) for r in rows)))

n = 101; x, u = fd(n); xm = 0.5 * (x[:-1] + x[1:])
print('\n节点误差最大 %+.6e ；单元中点误差最小 %+.6e  -> 反号，P1 重建抵消一部分节点误差'
      % ((u - exact(x))[np.argmax(np.abs(u - exact(x)))], (np.interp(xm, x, u) - exact(xm)).min()))
for tol, tag in [(1.096e-5, '软约束 1.096e-5'), (3.790e-6, '硬约束 3.790e-6')]:
    for name, idx in [('6.1口径', 3), ('6.4口径', 4)]:
        lo = [r for r in rows if r[idx] > tol]; hi = [r for r in rows if r[idx] <= tol]
        print('  %s  %s: 介于 N=%d(%.3e) 与 N=%d(%.3e) 之间'
              % (tag, name, lo[-1][0], lo[-1][idx], hi[0][0], hi[0][idx]))

print('\n反例常数：')
print('  ||1-x||/||sin|| = %.12f , sqrt(2/3) = %.12f' %
      (np.sqrt(np.dot(WQ, (1 - XQ) ** 2) / np.dot(WQ, exact(XQ) ** 2)), np.sqrt(2 / 3)))
print('  ||3x+2||/||sin|| = %.12f , sqrt(26) = %.12f' %
      (np.sqrt(np.dot(WQ, (3 * XQ + 2) ** 2) / np.dot(WQ, exact(XQ) ** 2)), np.sqrt(26)))
xx = np.linspace(0, 1, 2000001); ee = 3 * xx + 2 - np.sin(PI * xx)
print('  max|3x+2-sin| = %.6f @ x=%.1f ；min = %.6f @ x=%.5f' % (ee.max(), xx[ee.argmax()], ee.min(), xx[ee.argmin()]))
print('  (2.34) 与 (2.36) 在精确解族上的核验见单元 3；此处 5.0 = 端点，非「冒充」')
```

```python
# 单元 2 · 6.1 / A2：解析反例的真残差、自检断言是否可能失败、L_b 与端点
# 代码来源：AI 生成初稿；组员对照 A2组-实验.ipynb 单元 3 的 p61_counterexample 逐行核对
import numpy as np
PI = np.pi
x = np.linspace(0, 1, 1001)

def true_residual(v_xx, f):          # 用解析二阶导，避免有限差分在端点的一阶误差
    return -v_xx - f

f = PI ** 2 * np.sin(PI * x)
v_xx = -PI ** 2 * np.sin(PI * x)                     # v = sin(pi x)+3x+2 的解析二阶导
r_true = true_residual(v_xx, f)
print('A2 反例：真实残差 max|r| = %.3e （解析为 0）' % np.max(np.abs(r_true)))

# A2 代码写法：p61_residual = -(-pi^2 sin) - pi^2 sin
r_code = -(-PI ** 2 * np.sin(PI * x)) - PI ** 2 * np.sin(PI * x)
print('A2 代码式残差 max = %.3e  （算的是 sin(pi x) 的残差，与 v 无关）' % np.max(np.abs(r_code)))

# 故意破坏反例：v_bad = v + 0.1x^2 -> v_bad'' = -pi^2 sin + 0.2，真残差 = 0.2
v_xx_bad = -PI ** 2 * np.sin(PI * x) + 0.2
print('破坏后 v_bad 真实残差 max = %.6f' % np.max(np.abs(true_residual(v_xx_bad, f))),
      '；代码式仍为 %.1f' % np.max(np.abs(r_code)))
print('-> 该 assert 在任何输入下都通过，无法发现反例被改坏（结论本身仍正确）')

v = np.sin(PI * x) + 3 * x + 2
print('\n端点 (v(0),v(1)) = (%.1f, %.1f) ；L_b=(a^2+b^2)/2 = %.1f'
      % (v[0], v[-1], (v[0] ** 2 + v[-1] ** 2) / 2))
print('max|v-u*| = %.1f ；通解 sin(pi x)+ax+b 代入 x=0 得 b=0，代入 x=1 得 a=0' % np.max(np.abs(v - np.sin(PI * x))))
```

```python
# 单元 5 · 6.3 / A1：周期问题的解族维数（p 是否瓶颈）与 FNO 的两条结构事实
# 代码来源：AI 生成初稿；组员核对 -w''+w=u 的解与源项同频率这一推导
import numpy as np
PI = np.pi
KMAX, N, M = 4, 200, 1024
rng = np.random.default_rng(11)
xg = np.linspace(0, 1, M, endpoint=False)
U = np.zeros((N, M))
for k in range(KMAX + 1):
    U += rng.standard_normal((N, 1)) * np.cos(2 * PI * k * xg) + rng.standard_normal((N, 1)) * np.sin(2 * PI * k * xg)
lam = 1 + (2 * PI * np.fft.rfftfreq(M, d=1 / M)) ** 2
W = np.fft.irfft(np.fft.rfft(U, axis=1) / lam, n=M, axis=1)
for name, A in [('源项 u', U), ('解 w  ', W)]:
    sv = np.linalg.svd(A, compute_uv=False)
    print('%s 奇异值前 11: %s' % (name, np.array2string(sv[:11], precision=3)))
    print('        秩 = %d （理论 2K+1 = %d）, sv[9]/sv[0] = %.3e'
          % (int(np.sum(sv > 1e-10 * sv[0])), 2 * KMAX + 1, sv[9] / sv[0]))
print('-> p=32 >= 9：trunk 足够精确表示该解族，4.14e-3 的误差不是基的个数造成的')

n, kmax = 64, 8
x = np.arange(n) / n
Wr = rng.standard_normal(kmax) + 1j * rng.standard_normal(kmax)
def fno_layer(v, Wr=Wr):
    vh = np.fft.rfft(v); out = np.zeros_like(vh); out[:kmax] = vh[:kmax] * Wr
    return np.fft.irfft(out, n=n)
I = np.eye(n)
Mx = np.stack([fno_layer(I[:, j]) for j in range(n)], axis=1)
print('\nFNO 谱层算子矩阵的循环性 max|M[i,j]-M[0,(j-i) mod n]| = %.3e'
      % max(abs(Mx[i, j] - Mx[0, (j - i) % n]) for i in range(n) for j in range(n)))
v = np.exp(-((x - 0.05) / 0.02) ** 2)
out = fno_layer(v, Wr=np.ones(kmax, complex))
print('凸包支撑 [0,0.12] 经纯低通：原区 %.4f ；远区[0.3,0.7] %.4f ；缝合区[0.85,1) %.4f'
      % (np.max(np.abs(out - v)[:8]), np.max(np.abs(out)[19:45]), np.max(np.abs(out)[55:])))
print('每层实参 2*cin*cout*kmax, (32,32,8) =', 2 * 32 * 32 * 8)
```

```python
# 单元 6 · 6.3 / A2：解族维数 16（p 是否瓶颈）
# 代码来源：AI 生成初稿；组员核对 cos 项需减去常数以满足零边界
import numpy as np
PI = np.pi
rng = np.random.default_rng(7)
N, KMAX, M = 200, 8, 1024
kk = np.arange(1, KMAX + 1)
a = rng.standard_normal((N, KMAX)) * kk ** -1.5
b = rng.standard_normal((N, KMAX)) * kk ** -1.5
xg = np.linspace(0, 1, M)
u = np.zeros((N, M))
for j in range(KMAX):
    m = kk[j]
    u += (a[:, j, None] * np.cos(2 * PI * m * xg) + b[:, j, None] * np.sin(2 * PI * m * xg)) / (4 * PI ** 2 * m ** 2)
u = u - u[:, :1] + (u[:, -1:] - u[:, :1]) * xg[None, :]      # 加仿射项使 u(0)=u(1)=0
sv = np.linalg.svd(u, compute_uv=False)
print('奇异值前 20:', np.array2string(sv[:20], precision=3))
print('秩(阈值 1e-10*smax) = %d  sv[16]/sv[0] = %.3e' % (int(np.sum(sv > 1e-10 * sv[0])), sv[16] / sv[0]))
B = np.stack([np.cos(2 * PI * kk[j] * xg) - 1 for j in range(KMAX)] + [np.sin(2 * PI * kk[j] * xg) for j in range(KMAX)], axis=1)
coef, *_ = np.linalg.lstsq(B, u[0], rcond=None)
print('16 维解析基做最小二乘的最大残差 = %.3e' % np.max(np.abs(B @ coef - u[0])))
print('-> p=32 >= 16：削峰不能归因于基的个数；剩余误差在 branch 的回归与优化')
```

```python
# 单元 7 · 6.3 / A2 复算：逐字抄录 A2组-代码-6.3.py（仅删绘图段）
# 代码来源：**A2 组随稿脚本**（非 AI 生成）；本组只删 matplotlib 行并补末尾对照打印，训练/损失/误差口径未改一字
import time, numpy as np
np.random.seed(0)
t_gen = time.time()
M_FINE, M_COARSE = 1024, 256
x_f = (np.arange(1, M_FINE + 1)) / (M_FINE + 1)
x_c = (np.arange(1, M_COARSE + 1)) / (M_COARSE + 1)
def fd_inv(M):
    h = 1 / (M + 1)
    A = (2 * np.diag(np.ones(M)) - np.diag(np.ones(M - 1), 1) - np.diag(np.ones(M - 1), -1)) / h ** 2
    return np.linalg.inv(A)
Ainv_f, Ainv_c = fd_inv(M_FINE), fd_inv(M_COARSE)
KMAX, SENSOR = 8, 51
sensor_x = np.linspace(0, 1, SENSOR); _k = np.arange(1, KMAX + 1).astype(float); _dec = _k ** -1.5
def gen_coef(n): return np.random.randn(n, KMAX) * _dec, np.random.randn(n, KMAX) * _dec
def eval_f(a, b, x):
    X = x[None, :]; KK = _k[:, None]
    return (a[:, :, None] * np.cos(2 * np.pi * KK * X)[None] + b[:, :, None] * np.sin(2 * np.pi * KK * X)[None]).sum(1)
N_TR, N_TE = 800, 200
ar, br = gen_coef(N_TR); U_tr = (Ainv_c @ eval_f(ar, br, x_c).T).T; fs_tr = eval_f(ar, br, sensor_x)
ae, be = gen_coef(N_TE); U_te = (Ainv_f @ eval_f(ae, be, x_f).T).T; fs_te = eval_f(ae, be, sensor_x)
U_ct = (Ainv_c @ eval_f(ae, be, x_c).T).T
t_gen = time.time() - t_gen
P, H = 32, 64
def init(i, o): return np.random.randn(i, o) * np.sqrt(2 / i), np.zeros(o)
W1b, b1b = init(SENSOR, H); W2b, b2b = init(H, P); W1t, b1t = init(1, H); W2t, b2t = init(H, P)
pr = [W1b, b1b, W2b, b2b, W1t, b1t, W2t, b2t]
m = [np.zeros_like(p) for p in pr]; v = [np.zeros_like(p) for p in pr]
def fwd(fs, xq):
    hb = np.tanh(fs @ pr[0] + pr[1]); B = hb @ pr[2] + pr[3]
    ht = np.tanh(xq @ pr[4] + pr[5]); T = ht @ pr[6] + pr[7]
    return B, hb, T, ht, B @ T.T
x_tr = x_c[:, None]; U = U_tr; Xb = fs_tr
t0 = time.time()
for ep in range(1, 15001):
    B, hb, T, ht, Y = fwd(Xb, x_tr)
    D = (Y - U) / (U.shape[0] * U.shape[1])
    dT = D.T @ B; dht = dT @ pr[6].T * (1 - ht ** 2)
    gW2t, gB2t = ht.T @ dT, dT.sum(0); gW1t, gB1t = x_tr.T @ dht, dht.sum(0)
    dB = D @ T; dhb = dB @ pr[2].T * (1 - hb ** 2)
    gW2b, gB2b = hb.T @ dB, dB.sum(0); gW1b, gB1b = Xb.T @ dhb, dhb.sum(0)
    gs = [gW1b, gB1b, gW2b, gB2b, gW1t, gB1t, gW2t, gB2t]
    for i in range(8):
        m[i] = .9 * m[i] + .1 * gs[i]; v[i] = .999 * v[i] + .001 * gs[i] ** 2
        pr[i] -= 1e-3 * (m[i] / (1 - .9 ** ep)) / (np.sqrt(v[i] / (1 - .999 ** ep)) + 1e-8)
t_train = time.time() - t0
t0 = time.time(); _, _, _, _, U_pred = fwd(fs_te, x_f[:, None]); t_infer = time.time() - t0
print('train relL2=%.3e   (=||Y-U||_F/||U||_F，全局 Frobenius，不是同列逐函数口径)' % (np.linalg.norm(Y - U) / np.linalg.norm(U)))
print('time: data-gen=%.3e s/%d func | train=%.3e s | infer=%.3e ms/%d func'
      % (t_gen, N_TR + N_TE, t_train, t_infer * 1000, N_TE))
err_d = np.linalg.norm(U_pred - U_te, axis=1) / np.linalg.norm(U_te, axis=1)
U_cti = np.stack([np.interp(x_f, x_c, r) for r in U_ct])
err_fd = np.linalg.norm(U_cti - U_te, axis=1) / np.linalg.norm(U_te, axis=1)
print('DeepONet test relL2 mean=%.3e median=%.3e max=%.3e' % (err_d.mean(), np.median(err_d), err_d.max()))
print('FD coarse relL2 mean=%.3e median=%.3e max=%.3e' % (err_fd.mean(), np.median(err_fd), err_fd.max()))
print('\nA2 正文表6.3-4: 9.36e-2 | 1.60e-1 | 1.19e-1 | 1.20e+0 | 7.42e-4')
print('A2 保存输出:     1.661e-1 | 2.370e-1 | 1.978e-1')
print('B1 NumPy2.3.5:   9.699e-2 | 1.5186e-1 | 1.1283e-1 ; B1 NumPy1.26.4: 1.6612e-1 | 2.3705e-1 | 1.9784e-1')
print('B2 NumPy2.5.4:   9.359e-2 | 1.515e-1 | 1.133e-1 ; 本组 NumPy2.3.5 见上')
print('-> FD 基线 7.42e-4 四个来源全部命中（确定性）；训练/测试量跨环境漂移 1%-56%')
```

```python
# 单元 8 · 6.4 / A1：连续残差界 (6.4-11) 的独立重推与数值检验 + FEM 强残差不可用
# 代码来源：AI 生成初稿；组员独立重推 Poincare 两用与 ||l||^2=(a^2+ab+b^2)/3
import numpy as np
PI = np.pi
trap = getattr(np, 'trapezoid', np.trapz)
X = np.linspace(0, 1, 400001)

def check(v_fun, name):
    v = v_fun(X); a, b = v[0], v[-1]
    r = -np.gradient(np.gradient(v, X), X) - PI ** 2 * np.sin(PI * X)
    e = v - np.sin(PI * X)
    rhs = np.sqrt((a * a + a * b + b * b) / 3) + np.sqrt(trap(r ** 2, X)) / PI ** 2
    lhs = np.sqrt(trap(e ** 2, X))
    print('  %-26s a=%.3e b=%.3e  ||e||=%.6e  界=%.6e  成立=%s'
          % (name, a, b, lhs, rhs, rhs >= lhs - 1e-12))
    return lhs, rhs

print('A1 的 (6.4-11) 数值检验（界 = 边界项 + ||r||/pi^2）：')
l1, r1 = check(lambda z: np.sin(PI * z) * 1.002, 'v=1.002 sin（仅幅值偏）')
check(lambda z: np.sin(PI * z) + 0.3 * z + 0.1, 'v=sin+0.3x+0.1（边界偏）')
check(lambda z: np.sin(PI * z) * (1 + 0.05 * np.sin(3 * PI * z)), 'v=sin(1+0.05 sin3)')
check(lambda z: np.sin(PI * z) + 1e-3 * np.sin(5 * PI * z), 'v=sin+1e-3 sin5')
print('  第一例 %0.6e vs %0.6e：单模态扰动时边界项为 0 且 ||r||/pi^2 恰好等于 ||e||，界取等号（紧）' % (l1, r1))

# FEM：单元内 u_h 是直线 -> u_h''=0 -> 强残差 = -f
def tri(diag, off, rhs):
    m = len(rhs); A = np.zeros((m, m))
    A[np.arange(m), np.arange(m)] = diag
    A[np.arange(m-1), np.arange(1, m)] = off; A[np.arange(1, m), np.arange(m-1)] = off
    return np.linalg.solve(A, rhs)

def fem(n, order=4):
    x = np.linspace(0, 1, n+1); h = 1 / n
    z, w = np.polynomial.legendre.leggauss(order); t = (z + 1) / 2
    f = PI ** 2 * np.sin(PI * (x[:-1, None] + h * t[None, :]))
    load = np.zeros(n+1)
    load[:-1] += h/2*np.sum(f*(w*(1-t))[None, :], axis=1)
    load[1:] += h/2*np.sum(f*(w*t)[None, :], axis=1)
    diag = np.zeros(n+1); diag[:-1] += 1/h; diag[1:] += 1/h
    off = np.full(n, -1/h); u = np.zeros(n+1)
    u[1:-1] = tri(diag[1:-1], off[1:-1], load[1:-1])
    return x, u

x, uh = fem(100)
e = uh - np.sin(PI * x)
print('\nFEM(n=100)：节点误差 max = %.3e（节点精确）' % np.max(np.abs(e)))
print('  单元内强残差 -u_h\'\'-f = -f，max|.| = %.6f = pi^2' % PI ** 2)
print('  比值 = %.3e -> 用强残差判 FEM 会把一个精确方法判成「完全不收敛」' % (PI ** 2 / np.max(np.abs(e))))
```

```python
# 单元 9 · 6.4 / A2：式 (6.4-4) 的 pi^2 漏项、反例后果、命题双向检验、3 模态表示
# 代码来源：AI 生成初稿；组员核对 D_m 推导与反例构造 b_m = c_m/(k m^2)
import numpy as np
PI = np.pi
trap = getattr(np, 'trapezoid', np.trapz)
c = np.array([1.0, 0.6, -0.4])
k = 2.6
xm = np.linspace(0, 1, 40001)
B3 = np.stack([np.sin((m + 1) * PI * xm) for m in range(3)], axis=1)
u_true = B3 @ (c / (k * (np.arange(1, 4) * PI) ** 2))
f = B3 @ c
coef, *_ = np.linalg.lstsq(B3, u_true, rcond=None)
print('3 模态最小二乘重构残差 max|.| = %.3e （k=2.6 在外推区，表示仍精确）' % np.max(np.abs(B3 @ coef - u_true)))

b_bad = c / (k * np.arange(1, 4) ** 2)
r_text = B3 @ (k * np.arange(1, 4) ** 2 * b_bad - c)
u_hat = B3 @ b_bad; e = u_hat - u_true
r_true = -k * np.gradient(np.gradient(u_hat, xm), xm) - f
print('\n取 b_m = c_m/(k m^2)（按文中的「零残差解」）：')
print('  文中(6.4-4) 残差 max|r_text| = %.6f' % np.max(np.abs(r_text)))
print('  真实残差     max|r_true| = %.6f' % np.max(np.abs(r_true)))
print('  真实解误差 ||e||_2 = %.6f ；文中界右端 = %.6f （< 真误差，命题被自己的反例推翻）'
      % (np.sqrt(trap(e ** 2, xm)), np.sqrt(trap(r_text ** 2, xm)) / (k * PI ** 2)))
print('  正确界（用 D_m=km^2 pi^2 b_m - c_m）= %.6f ，成立：%s'
      % (np.sqrt(trap(r_true ** 2, xm)) / (k * PI ** 2),
         np.sqrt(trap(r_true ** 2, xm)) / (k * PI ** 2) >= np.sqrt(trap(e ** 2, xm))))
print('  相对解误差 = %.6f ；pi^2-1 = %.6f' % (np.sqrt(trap(e ** 2, xm) / trap(u_true ** 2, xm)), PI ** 2 - 1))

b_ok = c / (k * (np.arange(1, 4) * PI) ** 2) * (1 + 0.05 * np.array([1.0, -1.0, 1.0]))
u2 = B3 @ b_ok; e2 = u2 - u_true
r2 = -k * np.gradient(np.gradient(u2, xm), xm) - f
print('\n真解附近 5%% 系数扰动：||e||=%.6e , 界=%.6e , 成立：%s'
      % (np.sqrt(trap(e2 ** 2, xm)), np.sqrt(trap(r2 ** 2, xm)) / (k * PI ** 2),
         np.sqrt(trap(r2 ** 2, xm)) / (k * PI ** 2) >= np.sqrt(trap(e2 ** 2, xm))))
```
