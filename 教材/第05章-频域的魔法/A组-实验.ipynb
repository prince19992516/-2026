## 先看一幕：波形的分身术

两条波形来参加“身份识别”：一条跑得慢，一条跑得快。机器人只看八次快照，居然给它们贴上了同一个标签。是傅里叶变换失灵，还是快照没有记下全部信息？

**先猜，再验证：** 令 $x_j=2\pi j/8$，$j=0,\ldots,7$。为什么 $\sin(3x_j)$ 与 $\sin(11x_j)$ 一样？这些点相同，能否推出两个连续函数相同？

**把故事翻译成数学：** 因为 $11x_j=3x_j+2\pi j$，两组采样值完全相同，这是混叠的具体例子。图中用8点便于观察；本章起步代码用64点对比频率3与67，原理一致。魔法只是叙事，比对采样率与频率单位才是关键。

## 1. 学习目标
正确解释 FFT 频率单位；构造周期谱微分；识别混叠和周期假设。

## 2. 数学背景与适用范围
在 $[0,2\pi)$ 上对周期函数 $\sin(3x)$ 采样，解析导数为 $3\cos(3x)$。频域微分相当于把 Fourier 系数乘以 $ik$。NumPy 的 fftfreq 返回每单位采样长度的周数，需乘以 $2\pi$ 才得到角频率。

采样不包含重复的右端点。频率高于网格可分辨范围时可能混叠，不能把离散重构准确直接理解为连续函数恢复准确。

## 3. 实现与可执行断言
代码使用合成输入或解析函数，无需下载数据。断言检测上述具体示例，不构成一般正确性证明。


```python
import numpy as np

n = 64
length = 2*np.pi
x = np.linspace(0, length, n, endpoint=False)
u = np.sin(3*x)
k = 2*np.pi*np.fft.fftfreq(n, d=length/n)
du = np.fft.ifft(1j*k*np.fft.fft(u)).real
exact = 3*np.cos(3*x)
np.testing.assert_allclose(du, exact, atol=1e-12, rtol=0)
alias_signal = np.sin((n+3)*x)
np.testing.assert_allclose(alias_signal, u, atol=1e-12, rtol=0)
print({"derivative_max_error": float(np.max(np.abs(du-exact))),
       "alias_sample_error": float(np.max(np.abs(alias_signal-u)))})
```
