### 已有起步实验：先运行并解释

多项式在域内/域外误差的示例只说明分布改变的风险，不构成一般泛化界，也不是神经网络训练结果。

以下保留原有代码供阅读、修改和复现。它只是本节的一个基线；请继续完成下方空白区。

```python
import numpy as np
from numpy.polynomial import Polynomial

train_x = np.linspace(-1, 1, 41)
model = Polynomial.fit(train_x, np.sin(train_x), deg=3).convert()
inside_x = np.linspace(-0.975, 0.975, 40)
outside_x = np.linspace(2, 3, 101)
def rmse(x):
    return float(np.sqrt(np.mean((model(x)-np.sin(x))**2)))
train_error = rmse(train_x)
inside_error = rmse(inside_x)
outside_error = rmse(outside_x)
assert train_error < 1e-3
assert inside_error < 1e-3
assert outside_error > 100*inside_error
print({"train_rmse": train_error, "independent_in_domain_rmse": inside_error,
       "out_of_domain_rmse": outside_error})
```
