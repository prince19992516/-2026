### 已有起步实验：先运行并解释

一维Poisson差分制造解例子用作有限元的对照基线；它不是Euler/RK4实验，也不是已实现的有限元装配。

以下保留原有代码供阅读、修改和复现。它只是本节的一个基线；请继续完成下方空白区。

```python
import numpy as np
from scipy.linalg import solve_banded

def solve_poisson(n):
    x = np.arange(1, n + 1) / (n + 1)
    h = 1 / (n + 1)
    band = np.zeros((3, n))
    band[0, 1:] = -1 / h**2
    band[1, :] = 2 / h**2
    band[2, :-1] = -1 / h**2
    u = solve_banded((1, 1), band, np.pi**2 * np.sin(np.pi*x))
    return x, u

errors = []
for n in [15, 31, 63]:
    x, u = solve_poisson(n)
    errors.append(np.max(np.abs(u - np.sin(np.pi*x))))
orders = np.log2(np.array(errors[:-1]) / errors[1:])
assert np.all((orders > 1.95) & (orders < 2.05))
assert errors[-1] < 3e-4
print({"errors": errors, "orders": orders})
```
