## 1. 学习目标
从边值问题构造离散方程；检查边界条件与符号；通过制造解估计收敛阶。

## 2. 数学背景与适用范围
考虑 $-u^{\prime\prime}(x)=\pi^2\sin(\pi x)$，$x\in(0,1)$，且 $u(0)=u(1)=0$。解析解为 $u(x)=\sin(\pi x)$。内点中心差分给出 $(-u_{i-1}+2u_i-u_{i+1})/h^2=f_i$。

矩阵对角元为 $2/h^2$、相邻副对角为 $-1/h^2$。本例用三对角求解器验证二阶误差；有限元弱形式与装配由学生补充。残差小只表示离散方程求解充分，不保证离散模型或边界条件正确。

## 3. 实现与可执行断言
代码使用合成输入或解析函数，无需下载数据。断言检测上述具体示例，不构成一般正确性证明。


```python
import numpy as np
from scipy.linalg import solve_banded

def solve_poisson(n):
    x = np.arange(1, n + 1) / (n + 1)
    h = 1 / (n + 1)
    band = np.zeros((3, n))
    band[0, 1:] = -1 / h**2
    band[1, :] = 2 / h**2
    band[2, :-1] = -1 / h**2
    u = solve_banded((1, 1), band, np.pi**2 * np.sin(np.pi*x))
    return x, u

errors = []
for n in [15, 31, 63]:
    x, u = solve_poisson(n)
    errors.append(np.max(np.abs(u - np.sin(np.pi*x))))
orders = np.log2(np.array(errors[:-1]) / errors[1:])
assert np.all((orders > 1.95) & (orders < 2.05))
assert errors[-1] < 3e-4
print({"errors": errors, "orders": orders})
```
